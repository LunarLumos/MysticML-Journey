# 🚀 Getting Started with Jupyter Notebook
**Week 1 · Module 1 – Installation of Python on Anaconda Platform**

A notebook is made of **cells**:
* **Code cells** – run Python, output appears right below.
* **Markdown cells** – formatted text like this one (headings, lists, `code`, math $E = mc^2$).

Run a cell with **Shift + Enter**. Press **Esc** for *command mode*, **Enter** for *edit mode*.

## 1. Your first code cell

```python
print("Hello, Jupyter! 👋")
2 + 3   # the value of the LAST expression is displayed automatically
```

## 2. Variables live in the kernel
Variables defined in one cell are available in every cell run *after* it. The number in `In [n]` tells you the execution order.

```python
name = "MysticML learner"
week = 1
```

```python
print(f"Hi {name}, welcome to week {week}!")
```

## 3. Which Python am I using?

```python
import sys, platform
print("Python", platform.python_version())
print("Executable:", sys.executable)
```

## 4. Getting help
* `help(len)` or `len?` shows documentation
* **Shift + Tab** inside a function call shows its signature
* **Tab** autocompletes names

```python
help(len)
```

## 5. Magic commands (IPython only)
`%` = line magic, `%%` = cell magic.

```python
%timeit sum(range(1000))
```

```python
%who
```

## 6. Rich output
The last expression of a list/dict is pretty-printed.

```python
squares = {n: n ** 2 for n in range(1, 6)}
squares
```

## ✅ Useful keyboard shortcuts (command mode – press Esc first)
| Shortcut | Action |
|---|---|
| `A` / `B` | insert cell above / below |
| `D, D` | delete cell |
| `M` / `Y` | change cell to Markdown / Code |
| `Shift + Enter` | run cell, move to next |
| `Ctrl + Enter` | run cell, stay |
| `0, 0` | restart kernel |
